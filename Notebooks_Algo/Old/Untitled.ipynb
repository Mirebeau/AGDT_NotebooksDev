```python
@ti.pyfunc
def zero_like(x):
    y = x-x
    y.fill(0)
    return y

vec_t = ti.lang.matrix.VectorType(2,ti.i8) 
x = vec_t(20,30)
x0 = zero_like(x)
x0[0]+=40
print(x0,x)

@ti.kernel
def testzero():
    x = vec_t(20,30)
    x0 = zero_like(x)
    x0[1]+=40
    print(x0)
    print(x)
    x1 = ti.zero(x0)
    x1[0]+=120; x1[0]*=2
    print(x1)
testzero()
```

```python
@ti.pyfunc
def cast_vec(x,dtype:ti.template()):
    y = dtype(0)
    y = dtype(x)
    return y

@ti.pyfunc
def cast_scal(x,dtype:ti.template()):
    y:dtype = 0
    y = dtype(x)
    return y

vec = ti.math.vec2([2.4,120.5]); vec_t = ti.lang.matrix.VectorType(2,ti.i8) # vec_t = ti.math.ivec2
scal = 2.5; scal_t = ti.i32
#x = 2.5; dtype = ti.i32
py_iv = cast_vec(vec,vec_t)
py_iv[1]+=40
print("python",py_iv)
#print("python",cast_scal(scal,scal_t))

@ti.kernel
def testcast():
    iv = cast_vec(vec,vec_t)
    iv[1]+=40
    print(iv)
    print(cast_scal(scal,scal_t))
    print("ti cast",ti.cast(2.5,ti.i32))
testcast()
```
